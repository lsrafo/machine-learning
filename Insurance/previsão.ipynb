```python
import joblib
import pandas as pd
```

```python
modelo = joblib.load("regressao_linear_insurance.pkl")
scaler = joblib.load("scaler.pkl")
```

```python
def prever_valor(idade, indice_massa_corporal, criancas, fumante, sexo):

    colunas = [
        'age', 'bmi', 'children', 'smoker', 'female',
        'northwest', 'southeast', 'southwest'
    ]

    features = pd.DataFrame(
        [[idade, indice_massa_corporal, criancas, fumante, sexo, 1, 0, 0]],
        columns=colunas
    )

    features_escaladas = scaler.transform(features)

    # Mantém as 5 variáveis utilizadas na regressão
    features_modelo = pd.DataFrame(
        features_escaladas[:, :5],
        columns=modelo.feature_names_in_
    )

    previsao = modelo.predict(features_modelo)

    return float(previsao.item())
```

```python
valor = prever_valor(
    idade=30,
    indice_massa_corporal=25.5,
    criancas=2,
    fumante=0,
    sexo=0
)

print(f"Preço previsto do seguro: R$ {valor:.2f}")
```
