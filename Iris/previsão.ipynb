```python
import joblib
import pandas as pd
```

```python
modelo = joblib.load("knn_iris.pkl")
```

```python
nomes = {
    0: "setosa",
    1: "versicolor",
    2: "virginica"
}
```

```python
def prever_flor(sepal_length, sepal_width, petal_length, petal_width):

    nova_flor = pd.DataFrame(
        [[
            sepal_length,
            sepal_width,
            petal_length,
            petal_width
        ]],
        columns=[
            "sepal length (cm)",
            "sepal width (cm)",
            "petal length (cm)",
            "petal width (cm)"
        ]
    )

    previsao = modelo.predict(nova_flor)

    print(f"Espécie prevista: {nomes[previsao[0]]}")
```

```python
prever_flor(5.1, 3.5, 1.4, 0.2)
```
