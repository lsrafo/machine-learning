```python
import joblib
```

```python
modelo = joblib.load("knn_iris.pkl")
```

```python
nomes = {
    0: "setosa",
    1: "versicolor",
    2: "virginica"
}
```

```python
sepal_length = float(input("Sepal length: "))
sepal_width = float(input("Sepal width: "))
petal_length = float(input("Petal length: "))
petal_width = float(input("Petal width: "))
```

```python
nova_flor = [[
    sepal_length,
    sepal_width,
    petal_length,
    petal_width
]]

previsao = modelo.predict(nova_flor)

print(f"Espécie prevista: {nomes[previsao[0]]}")
```
